```python
import pandas as pd

import zipfile
from pathlib import Path
```

```python
project_path = Path("..")
zip_path = project_path / "ecommerce_dataset.zip"
raw_path = project_path / "data" / "raw"
```

```python
raw_path.mkdir(parents = True, exist_ok = True)
```

```python
type(raw_path)
```

```python
with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(raw_path)
```

```python
list(raw_path.iterdir())
```

```python
csv_files = list(raw_path.glob("*.csv"))
csv_files
```

```python
dataframes = {}

for file in csv_files:
    dataframes[file.stem] = pd.read_csv(file)
```

```python
dataframes.keys()
```

```python
for name, df in dataframes.items():
    print(name, df.shape)
```

```python
for name, df in dataframes.items():
    print(f"\n{name.upper()}")
    print(df.columns.tolist())
```

```python
for name, df in dataframes.items():
    print(f"\n{name.upper()}")
    display(df.head())
```

```python
for name, df in dataframes.items():
    print(f"\n{name.upper()}")
    df.info()
```

```python
for name, df in dataframes.items():
    print(f"\n{name.upper()}")
    print(df.isna().sum())
```

```python
for name, df in dataframes.items():
    print(f"{name}:{df.duplicated().sum()} duplicate rows")
```

```python
for name, df in dataframes.items():
    print(f"{name.upper()}")
    print(df.dtypes)
```

```python
for name, df in dataframes.items():
    print(f"{name.upper()}")
    print(df.nunique())
```
